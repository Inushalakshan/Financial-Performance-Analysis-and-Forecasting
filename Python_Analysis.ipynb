```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Data_set.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
financial_columns = [
    'Revenue',
    'COGS',
    'Operating_Expenses',
    'Marketing_Expenses',
    'Depreciation_Amortization',
    'Interest_Expense',
    'Tax_Rate'
]

for col in financial_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df['Gross_Profit'] = df['Revenue']-df['COGS']
```

```python
df['EBITDA'] = df['Revenue']-df['COGS']-df['Operating_Expenses']-df['Marketing_Expenses']
```

```python
df['EBIT']= df['EBITDA']-df['Depreciation_Amortization']
```

```python
df['PBT']= df['EBIT']-df['Interest_Expense']
```

```python
df['Tax_Expense'] = (
    df['PBT'] * df['Tax_Rate']
)
```

```python
df['Net_Profit']=df['PBT']-df['Tax_Expense']
```

```python
df.head()
```

```python
# Calculate financial ratios
df['Gross_Margin'] = df['Gross_Profit'] / df['Revenue']
df['EBITDA_Margin'] = df['EBITDA'] / df['Revenue']
df['EBIT_Margin'] = df['EBIT'] / df['Revenue']
df['Net_Profit_Margin'] = df['Net_Profit'] / df['Revenue']
df['Interest_Coverage'] = df['EBIT'] / df['Interest_Expense']

df.head()
```

```python
df['Revenue_Growth'] = (
    df['Revenue'].pct_change() * 100
)

df['Profit_Growth'] = (
    df['Net_Profit'].pct_change() * 100
)

df['EBITDA_Growth'] = (
    df['EBITDA'].pct_change() * 100
)

print(df[[
    'Year',
    'Revenue_Growth',
    'Profit_Growth',
    'EBITDA_Growth'
]])
```

```python
revenue_cagr = (
    (df['Revenue'].iloc[-1] /
     df['Revenue'].iloc[0]) ** (1/4)
) - 1

print("Revenue CAGR:", round(revenue_cagr * 100, 2), "%")
```

```python
profit_cagr = (
    (df['Net_Profit'].iloc[-1] /
     df['Net_Profit'].iloc[0]) ** (1/4)
) - 1

print("Profit CAGR:", round(profit_cagr * 100, 2), "%")
```

```python
COGS_cagr = (
    (df['COGS'].iloc[-1] /
     df['COGS'].iloc[0]) ** (1/4)
) - 1

print("COGS CAGR:", round(COGS_cagr * 100, 2), "%")
```

```python
latest = df.iloc[-1]

print("Revenue-2027: ", latest['Revenue'] * (1 + revenue_cagr))
print("Profit-2027: ", latest['Net_Profit'] * (1 + profit_cagr))
print("COGS-2027: ", latest['COGS'] * (1 + COGS_cagr))
```
